```python
#4.4.2.gc_contents.py

from Bio.Seq import Seq

exon_seq = Seq("ATGCAGTAG")
g_count = exon_seq.count("G")
c_count = exon_seq.count("C")
gc_contents = (g_count + c_count) / len(exon_seq) * 100
print(gc_contents)
```

```python
#4.4.4.translate_transcribe.py

from Bio.Seq import Seq

dna = Seq("ATGCAGTAG")
mrna = dna.transcribe()
ptn = dna.translate()
print(mrna)
print(ptn) # *기호는 단백질 번역 과정이 끝나는 종결 코돈
```

```python
#4.4.5.translates_stop_1.py

from Bio.Seq import Seq

mrna = Seq("AUGAACUAAGUUUAGAAU")
ptn = mrna.translate()
print(ptn) #중간에 종결 코돈 삽입됨
```

```python
#4.4.5.translates_stop_2.py
#처음 종결 코돈에서 번역을 종료하려면?

from Bio.Seq import Seq

mRNA = Seq("AUGAACUAAGUUUAGAAU")
ptn = mRNA.translate(to_stop=True)
print(ptn) #중간에 종결 코돈 삽입됨
```

```python
#4.4.6.split.py
#종결 코돈 기준으로 서열 나누기

from Bio.Seq import Seq

mrna = Seq("AUGAACUAAGUUUAGAAU")
ptn = mrna.translate()
print(ptn) ## MN*V*N
for seq in ptn.split('*'):
    print(seq)
```

```python
#4.4.7.complement.py

seq = "TATAAAGGCAATATGCAGTAG"
comp_dic = {'A':'T', 'C':'G', 'G':'C','T':'A'} #상보적 염기를 키-값으로 하는 사전 생성
comp_seq = ""
for s in seq : #서열에서 하나씩 읽기
    comp_seq += comp_dic[s] #상보적 염기 추가
revcomp_seq = comp_seq[::-1] #문자열 뒤집기
print(comp_seq)
print(revcomp_seq)
```

```python
#4.4.7.complement.bio.py

from Bio.Seq import Seq

seq = Seq("TATAAAGGCAATATGCAGTAG")
comp_seq = seq.complement()
rev_comp_seq = seq.reverse_complement()
print(comp_seq)
print(rev_comp_seq)
```

```python
#4.4.9.orf_finder.py

from Bio.Seq import Seq

tatabox_seq = Seq("tataaaggcAATATGCAGTAG")
start_idx = tatabox_seq.find("ATG")
end_idx = tatabox_seq.find("TAG",start_idx)
orf = tatabox_seq[start_idx:end_idx+3]
print(orf)
```

```python
from Bio.Seq import Seq
# from Bio.Alphabet import IUPAC # 삭제
from Bio.SeqUtils import molecular_weight

seq1 = Seq("ATGCAGTAG")
# seq2 = Seq("ATGCAGTAG", IUPAC.unambiguous_dna) # 삭제
# seq3 = Seq("ATGCAGTAG", IUPAC.protein) # 삭제

print(molecular_weight(seq1))   # 기본적으로 sequence type은 DNA 입니다.
# print(molecular_weight(seq2)) # 삭제
# print(molecular_weight(seq3)) # 삭제

print(molecular_weight(seq1, seq_type="DNA")) # seq_type="DNA"로 DNA를 명시적으로 설정하였습니다.
print(molecular_weight(seq1, seq_type="protein")) # seq_type="protein" 으로 설정하였습니다.

# 위 스크립트를 실행하면
# 2842.82
# 2842.82
# 707.75
# 가 출력됩니다.

#서열은 같지만 DNA와 단백질의 서열의 차이 때문에 무게가 다르다!
#A = Alanine
#T = Threonine
#G = Glycine
#C = Cysteine
```
