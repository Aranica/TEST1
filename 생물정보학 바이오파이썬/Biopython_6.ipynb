보통 연구/업무에서 다루는 FASTQ 파일은 압축된 형태
-> format은 GNU Zip (gzip) 파일 형태
파이썬의 gzip module을 활용하면 압축된 FASTQ파일은 fastq.gz 파일을 읽을 수 있음

```python
#6.2.2.read_example_1.py

import gzip
from Bio import SeqIO

handle = gzip.open("sample_1.fastq.gz","rt")
seq = SeqIO.parse(handle, "fastq")
for s in seq:
    print(s.seq)
```

```python
#6.2.2.read_example_2.py

import gzip
from Bio import SeqIO

with gzip.open("sample_1.fastq.gz","rt") as handle:
    seq = SeqIO.parse(handle, "fastq")
    for s in seq:
        print(s.seq)
```

Biopython의 Entrez module 활용 -> 인터넷 상의 데이터를 받아와 SeqIO 모듈로 읽는 방법

Entrez 모듈 : NCBI 정보를 웹을 통해 접근하여 정보를 가져오는 모듈
             작업자의 이메일이 의무적으로 들어감
            Entrez.efetch() 메서드에 키워드 인자로 db,rettype,id,retmode 작성
                                            (데이터베이스 이름, 데이터 형식, 아이디, 반환 형식)

```python
#변이가 없는 CCR5 유전자에 대해 GenBank 형식으로 받아오는 방법
#6.4.2.entrez_example.py

from Bio import Entrez

Entrez.email = "jeanworldly@gmail.com"
handle = Entrez.efetch(db="nucleotide",rettype="gb",id="AY463215",retmode="text")

for s in handle:
    print(s.strip())
```

```python
#6.4.3.read_example_GenBank_1.py

from Bio import Entrez
from Bio import SeqIO

Entrez.email = "jeanworldly@gmail.com"

with Entrez.efetch(db="nucleotide",rettype="fasta",retmode="text",id="42540826") as handle:
    seq = SeqIO.read(handle,"fasta")

print(seq)
print(len(seq))
```

```python
#변이가 있는 CCR5 유전자에 대해 GenBank 형식으로 받아오는 방법

#6.4.3.read_example_GenBank_2.py

from Bio import Entrez
from Bio import SeqIO

Entrez.email = "jeanworldly@gmail.com"

with Entrez.efetch(db="nucleotide",rettype="fasta",retmode="text",id="1575550") as handle:
    seq = SeqIO.read(handle,"fasta")

print(seq)
print(len(seq))
```
